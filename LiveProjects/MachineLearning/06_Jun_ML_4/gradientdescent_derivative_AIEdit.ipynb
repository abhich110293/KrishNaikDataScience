For a linear regression model with a bias term, two features, and a mean squared error (MSE) loss function, here is the breakdown of the loss function derivatives in both scalar and vector form, along with the weight update rules.

### 1. Mathematical Setup

Given a dataset with $N$ samples, $2$ features ($x_1, x_2$), and target $y$:

- **Prediction for sample $i$:** $\hat{y}^{(i)} = w_1 x_{i1} + w_2 x_{i2} + b$
- **Revised MSE Loss Function ($L$):** $$ L(w_1, w_2, b) = \frac{1}{2N} \sum_{i=1}^{N} \left(\hat{y}^{(i)} - y^{(i)}\right)^2 = \frac{1}{2N} \sum_{i=1}^{N} \left(w_1 x_{i1} + w_2 x_{i2} + b - y^{(i)}\right)^2 $$ *(Note: The factor of $\frac{1}{2}$ is the standard "revised" convention used to cancel out the power of $2$ when taking the derivative).*

### 2. Scalar Calculus Formulation

To compute the gradients, we apply the chain rule with respect to each parameter ($w_1$, $w_2$, and $b$).

#### Derivative with respect to $w_1$:

$$
\frac{\partial L}{\partial w_1} = \frac{\partial}{\partial w_1} \left[ \frac{1}{2N} \sum_{i=1}^{N} \left(\hat{y}^{(i)} - y^{(i)}\right)^2 \right]
$$

$$
\frac{\partial L}{\partial w_1} = \frac{1}{2N} \sum_{i=1}^{N} 2 \cdot \left(\hat{y}^{(i)} - y^{(i)}\right) \cdot \frac{\partial}{\partial w_1}\left(w_1 x_{i1} + w_2 x_{i2} + b - y^{(i)}\right)
$$

$$
\frac{\partial L}{\partial w_1} = \frac{1}{N} \sum_{i=1}^{N} \left(\hat{y}^{(i)} - y^{(i)}\right) x_{i1}
$$

#### Derivative with respect to $w_2$:

$$
\frac{\partial L}{\partial w_2} = \frac{1}{N} \sum_{i=1}^{N} \left(\hat{y}^{(i)} - y^{(i)}\right) x_{i2}
$$

#### Derivative with respect to bias $b$:

$$
\frac{\partial L}{\partial b} = \frac{1}{N} \sum_{i=1}^{N} \left(\hat{y}^{(i)} - y^{(i)}\right)
$$

### 3. Vector Calculus Formulation

We group the features and weights using matrix notation by incorporating the bias into the weight vector:

- **Design Matrix ($X$):** $N \times 3$ matrix where the first column is $1$ (for bias): $$ X = \begin{bmatrix} 1 & x_{11} & x_{12} \\ 1 & x_{21} & x_{22} \\ \vdots & \vdots & \vdots \\ 1 & x_{N1} & x_{N2} \end{bmatrix} $$ 


- **Weight Vector ($\mathbf{w}$):** $3 \times 1$ vector where $\mathbf{w} = \begin{bmatrix} b & w_1 & w_2 \end{bmatrix}^T$
- **Target Vector ($\mathbf{y}$):** $N \times 1$ vector of true targets.
- **Prediction Vector ($\hat{\mathbf{y}}$):** $\hat{\mathbf{y}} = X \mathbf{w}$


#### Loss Function in Vector Form:
$$

L(\mathbf{w}) = \frac{1}{2N} (\mathbf{\hat{y}} - \mathbf{y})^T (\mathbf{\hat{y}} - \mathbf{y}) = \frac{1}{2N} (X\mathbf{w} - \mathbf{y})^T (X\mathbf{w} - \mathbf{y})

$$
Expanding the matrix product:
$$

L(\mathbf{w}) = \frac{1}{2N} \left( \mathbf{w}^T X^T X \mathbf{w} - 2 \mathbf{w}^T X^T \mathbf{y} + \mathbf{y}^T \mathbf{y} \right)

$$
#### Derivative with respect to Vector $\mathbf{w}$:

Using matrix calculus identities ($\nabla_{\mathbf{w}} (\mathbf{w}^T A \mathbf{w}) = 2 A \mathbf{w}$ and $\nabla_{\mathbf{w}} (\mathbf{w}^T A) = A$):
$$

\nabla_{\mathbf{w}} L(\mathbf{w}) = \frac{\partial L}{\partial \mathbf{w}} = \frac{1}{2N} \left( 2 X^T X \mathbf{w} - 2 X^T \mathbf{y} \right)

$$
$$
\nabla_{\mathbf{w}} L(\mathbf{w}) = \frac{1}{N} X^T (X\mathbf{w} - \mathbf{y}) = \frac{1}{N} X^T (\hat{\mathbf{y}} - \mathbf{y})
$$

### 4. Weight Updates (Gradient Descent)

Using a learning rate $\alpha \gt 0$, parameter updates move in the direction opposite to the gradient to minimize loss.

#### Scalar Update Rule:

For each iteration step:

$$
w_1^{(t+1)} = w_1^{(t)} - \alpha \cdot \frac{1}{N} \sum_{i=1}^{N} \left(\hat{y}^{(i)} - y^{(i)}\right) x_{i1}
$$

$$
w_2^{(t+1)} = w_2^{(t)} - \alpha \cdot \frac{1}{N} \sum_{i=1}^{N} \left(\hat{y}^{(i)} - y^{(i)}\right) x_{i2}
$$

$$
b^{(t+1)} = b^{(t)} - \alpha \cdot \frac{1}{N} \sum_{i=1}^{N} \left(\hat{y}^{(i)} - y^{(i)}\right)
$$

#### Vector Update Rule:

In compact vector notation, all parameters are updated simultaneously:

$$
\mathbf{w}^{(t+1)} = \mathbf{w}^{(t)} - \alpha \nabla_{\mathbf{w}} L(\mathbf{w}^{(t)})
$$

$$
\mathbf{w}^{(t+1)} = \mathbf{w}^{(t)} - \frac{\alpha}{N} X^T (X\mathbf{w}^{(t)} - \mathbf{y})
$$
