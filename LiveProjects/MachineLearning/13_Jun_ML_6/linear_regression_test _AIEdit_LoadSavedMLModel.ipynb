```python
# Import joblib for loading saved machine learning models and preprocessors
# joblib is efficient for serializing and deserializing large NumPy arrays and sklearn objects
# Import pandas for creating DataFrames, which is required for preprocessing input features
import joblib
import pandas as pd
```

```python
# Store the file paths to the saved model and preprocessor artifacts
# These files were created during the training phase and stored as joblib format
model_filename = "artefacts/final_sgd_reg_model.joblib"
preprocessor_filename = "artefacts/data_preprocessor.joblib"

# Load the saved SGD regression model from disk
# joblib.load reconstructs the trained model object with all learned coefficients and parameters
loaded_model = joblib.load(model_filename)
# Load the saved preprocessing pipeline from disk
# This preprocessor contains the fitted scalers and encoders that transform raw features
loaded_preprocessor = joblib.load(preprocessor_filename)
```

```python
# Define the list of feature column names that the model expects
# These names must match the order and names used during model training
# They correspond to Boston housing dataset attributes
FEATURE_COLUMNS = [
    'crim', 'zn', 'indus', 'chas', 'nox', 'rm', 'age', 
    'dis', 'rad', 'tax', 'ptratio', 'black', 'lstat'
]
```

```python
# Define a reusable function to make predictions from a list of feature values
def make_prediction_from_list(input_list, feature_names, loaded_model, loaded_preprocessor):
    # Convert the input list into a pandas DataFrame with proper column names
    # A DataFrame is required by the preprocessing pipeline
    X_input = pd.DataFrame([input_list], columns=feature_names)
    # print(f"\nInput DataFrame:\n{X_input}")

    # Transform the input features using the loaded preprocessor
    # Use transform() only (not fit_transform()) because the preprocessor was already fitted during training
    # Fitting on new data would alter the scaling/encoding rules, causing prediction errors
    X_input_processed = loaded_preprocessor.transform(X_input) #no fit required, just transform because the preprocessor is already fitted
    # Use the loaded model to generate predictions on the preprocessed features
    # The model returns an array of predictions; we take the first (and only) element
    predictions = loaded_model.predict(X_input_processed)
    # print(predictions[0])
    # Return the predicted value as a scalar
    return predictions[0]
```

```python
# Define a test case with sample feature values for prediction
# The order of values must match the order defined in FEATURE_COLUMNS
test_features_list = [
    5,    # crim (per capita crime rate)
    0,     # zn (proportion of residential land zoned for large lots)
    18.1,  # indus (proportion of non-retail business acres)
    0,     # chas (binary: 1 if property bounds Charles River, 0 otherwise)
    0.7,   # nox (nitric oxide concentration)
    4.0,   # rm (average number of rooms per dwelling)
    100,   # age (proportion of owner-occupied units built before 1940)
    1.0,   # dis (weighted distances to employment centers)
    24,    # rad (index of accessibility to radial highways)
    666,   # tax (property tax rate)
    20.2,  # ptratio (pupil-teacher ratio by town)
    300,   # black (1000*(Bk - 0.63)^2, where Bk is proportion of Black residents)
    30     # lstat (percentage of lower status population)
]
```

```python
# Call the prediction function with the test features, loaded model, and loaded preprocessor
# This returns the predicted median home value for the given feature values
predicted_value = make_prediction_from_list(test_features_list, FEATURE_COLUMNS, loaded_model, loaded_preprocessor)
```

```python
# Display the predicted median home value formatted to 2 decimal places
# The prediction is in thousands of dollars (medv unit from the Boston housing dataset)
print(f"Predicted medv: {predicted_value:.2f}")
```
