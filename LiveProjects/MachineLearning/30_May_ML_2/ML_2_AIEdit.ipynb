## **Lecture Summary: Machine Learning - Part 2**

This session covers fundamental concepts required before training algorithms, focusing on **Evaluation Metrics, Loss & Cost Functions, Model Fit Scenarios (Overfitting/Underfitting), and Regression Metrics**.

### **1. Core Concepts: Loss Function vs. Cost Function**

- **Loss Function**: Measures prediction error for a **single data point** ($y - \hat{y}$) .
- **Cost Function**: The **sum/average of loss functions** across an entire dataset (Train, Validation, or Test) .
- **Purpose**:
- *Loss Function*: Helps the model adjust itself during training (learning process) .
- *Evaluation Metrics*: Evaluates overall performance on test/validation data .

### **2. Model Performance Scenarios**

| Scenario | Train Loss / Error | Test / Validation Loss | Description |
| --- | --- | --- | --- |
| **Overfitting** | Very Low (e.g., $180$) | Very High (e.g., $400$) | Model memorizes training data too well; fails to generalize to unseen test data . |
| **Underfitting** | High (e.g., $250$) | High (e.g., $230$) | Model fails to learn underlying data patterns on both sets . |
| **Ideal / Good Fit** | Low (e.g., $30$) | Low / Close (e.g., $32$) | Model generalizes well to unseen data . |

### **3. Regression Evaluation Metrics**

#### **Key Formulas**

1. **Simple Difference**: $$ \sum_{i=1}^{n} (y - \hat{y}) $$
2. **Mean Squared Error (MSE)**: $$ \text{MSE} = \frac{1}{n}\sum_{i=1}^{n}(y - \hat{y})^2 $$
- *Advantage*: Penalizes larger errors more heavily due to squaring .
- *Disadvantage*: Changes the unit of measurement to target variable squared .
3. **Mean Absolute Error (MAE)**: $$ \text{MAE} = \frac{1}{n}\sum_{i=1}^{n}\vert{}y - \hat{y}\vert{} $$
- *Advantage*: Maintains the **same unit as the target variable**, making it easily interpretable .
4. **Root Mean Squared Error (RMSE)**: $$ \text{RMSE} = \sqrt{\frac{1}{n}\sum_{i=1}^{n}(y - \hat{y})^2} $$
- *Advantage*: Penalizes larger errors while maintaining the original target variable's unit scale .
5. **Modified MSE (Cost Function for Optimization)**: $$ \text{Modified MSE} = \frac{1}{2n}\sum_{i=1}^{n}(y - \hat{y})^2 $$
- *Purpose*: Adding the factor of $\frac{1}{2}$ simplifies calculations when taking the **derivative** $\frac{\partial}{\partial \theta}$ during gradient descent, as the power $2$ cancels out with $\frac{1}{2}$ .

### **4. Classification Evaluation Metrics (Confusion Matrix)**

- **Confusion Matrix Components**:
- **True Positive (TP)**: Truly classified as Positive .
- **True Negative (TN)**: Truly classified as Negative .
- **False Positive (FP)**: Falsely classified as Positive .
- **False Negative (FN)**: Falsely classified as Negative .
- **Accuracy**: $$ \text{Accuracy} = \frac{\text{TP} + \text{TN}}{\text{TP} + \text{TN} + \text{FP} + \text{FN}} $$
- Measures overall correctness .
- **Precision**: $$ \text{Precision} = \frac{\text{TP}}{\text{TP} + \text{FP}} $$
- Measures how often predicted positive cases are actually correct .
- **Recall**: $$ \text{Recall} = \frac{\text{TP}}{\text{TP} + \text{FN}} $$
- Measures how many actual positive cases were correctly caught .
- **F1-Score**: $$ \text{F1-Score} = \frac{2 \times (\text{Precision} \times \text{Recall})}{\text{Precision} + \text{Recall}} $$
- Harmonic mean of Precision and Recall .
