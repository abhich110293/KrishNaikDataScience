## **Lecture Summary: Machine Learning - Part 2**

This session covers fundamental concepts required before training algorithms, focusing on **Evaluation Metrics, Loss & Cost Functions, Model Fit Scenarios (Overfitting/Underfitting), and Regression Metrics**.

### **1. Core Concepts: Loss Function vs. Cost Function**

- **Loss Function**: Measures prediction error for a **single data point** ($y - \hat{y}$) .
- **Cost Function**: The **sum/average of loss functions** across an entire dataset (Train, Validation, or Test) .
- **Purpose**:
- *Loss Function*: Helps the model adjust itself during training (learning process) .
- *Evaluation Metrics*: Evaluates overall performance on test/validation data .

### **2. Model Performance Scenarios**

| Scenario | Train Loss / Error | Test / Validation Loss | Description |
| --- | --- | --- | --- |
| **Overfitting** | Very Low (e.g., $180$) | Very High (e.g., $400$) | Model memorizes training data too well; fails to generalize to unseen test data . |
| **Underfitting** | High (e.g., $250$) | High (e.g., $230$) | Model fails to learn underlying data patterns on both sets . |
| **Ideal / Good Fit** | Low (e.g., $30$) | Low / Close (e.g., $32$) | Model generalizes well to unseen data . |

### **3. Regression Evaluation Metrics**

#### **Key Formulas**

1. **Simple Difference**: $$ \sum_{i=1}^{n} (y - \hat{y}) $$
2. **Mean Squared Error (MSE)**: $$ \text{MSE} = \frac{1}{n}\sum_{i=1}^{n}(y - \hat{y})^2 $$
- *Advantage*: Penalizes larger errors more heavily due to squaring .
- *Disadvantage*: Changes the unit of measurement to target variable squared .
3. **Mean Absolute Error (MAE)**: $$ \text{MAE} = \frac{1}{n}\sum_{i=1}^{n}\vert{}y - \hat{y}\vert{} $$
- *Advantage*: Maintains the **same unit as the target variable**, making it easily interpretable .
4. **Root Mean Squared Error (RMSE)**: $$ \text{RMSE} = \sqrt{\frac{1}{n}\sum_{i=1}^{n}(y - \hat{y})^2} $$
- *Advantage*: Penalizes larger errors while maintaining the original target variable's unit scale .
5. **Modified MSE (Cost Function for Optimization)**: $$ \text{Modified MSE} = \frac{1}{2n}\sum_{i=1}^{n}(y - \hat{y})^2 $$
- *Purpose*: Adding the factor of $\frac{1}{2}$ simplifies calculations when taking the **derivative** $\frac{\partial}{\partial \theta}$ during gradient descent, as the power $2$ cancels out with $\frac{1}{2}$ .

### **4. Classification Evaluation Metrics (Confusion Matrix)**

- **Confusion Matrix Components**:
- **True Positive (TP)**: Truly classified as Positive .
- **True Negative (TN)**: Truly classified as Negative .
- **False Positive (FP)**: Falsely classified as Positive .
- **False Negative (FN)**: Falsely classified as Negative .
- **Accuracy**: $$ \text{Accuracy} = \frac{\text{TP} + \text{TN}}{\text{TP} + \text{TN} + \text{FP} + \text{FN}} $$
- Measures overall correctness .
- **Precision**: $$ \text{Precision} = \frac{\text{TP}}{\text{TP} + \text{FP}} $$
- Measures how often predicted positive cases are actually correct .
- **Recall**: $$ \text{Recall} = \frac{\text{TP}}{\text{TP} + \text{FN}} $$
- Measures how many actual positive cases were correctly caught .
- **F1-Score**: $$ \text{F1-Score} = \frac{2 \times (\text{Precision} \times \text{Recall})}{\text{Precision} + \text{Recall}} $$
- Harmonic mean of Precision and Recall .

To build intuition for the metrics derived from a Confusion Matrix, think of them as answers to **different real-world questions** you might ask a detector or an expert witness.

## 1. Precision — "The Quality Check"

$$
\text{Precision} = \frac{\text{True Positives}}{\text{True Positives} + \text{False Positives}}
$$

- **Question it answers:** *"When the model claims something is a Positive, how often is it actually right?"*
- **Where it looks:** Focuses **only on the model's positive predictions** (the top row of the matrix).
- **Everyday Analogy:** Imagine a fisherman who drops a net and pulls up 10 fish, but 3 of them are discarded plastic boots. His **Precision** is $7/10 = 70\%$.
- **When high Precision is critical:**
- **YouTube Video Recommendations:** If YouTube recommends 5 videos, you want all 5 to be relevant. If 4 are irrelevant spam, you stop using the app.
- **Stock Buy Signals:** If an algorithm tells you to buy a stock, you want high certainty it will go up so you don't lose capital.

## 2. Recall (Sensitivity or TPR) — "The Thoroughness Check"

$$
\text{Recall} = \frac{\text{True Positives}}{\text{True Positives} + \text{False Negatives}}
$$

- **Question it answers:** *"Out of all the real-world Positives that actually existed, how many did the model manage to catch?"*
- **Where it looks:** Focuses **only on actual real-world positive cases** (the left column of the matrix).
- **Everyday Analogy:** Imagine a lake has 100 total fish. A fisherman's net catches 20 of them. His **Recall** is $20/100 = 20\%$, regardless of how many plastic boots he pulled up.
- **When high Recall is critical:**
- **Fraud Detection:** Banks want to catch 99% of stolen credit card transactions, even if it means occasionally sending an annoying security SMS to an innocent customer.
- **Earthquake Warning Systems:** You want the alarm to trigger for every real earthquake.

## 3. Specificity (True Negative Rate) — "The Noise Rejection"

$$
\text{Specificity} = \frac{\text{True Negatives}}{\text{True Negatives} + \text{False Positives}}
$$

- **Question it answers:** *"Out of all the completely normal/safe cases, how good is the model at leaving them alone?"*
- **Where it looks:** Focuses **only on actual real-world negative cases** (the right column of the matrix).
- **Intuition:** This is the direct mirror image of Recall. While Recall measures how well you catch the sickness, Specificity measures how well you clear the healthy.
- **When high Specificity is critical:**
- **Medical Screening before Invasive Surgery:** Before performing a risky surgery, you need to be extremely confident the patient is actually sick (high Specificity) so you don't operate on a healthy person.

## 4. F1-Score — "The Balanced Grade"

$$
\text{F1-Score} = 2 \cdot \frac{\text{Precision} \cdot \text{Recall}}{\text{Precision} + \text{Recall}}
$$

- **Question it answers:** *"What is the overall balance between Precision and Recall?"*
- **Why not a simple average?** The F1-score uses a **Harmonic Mean**, which heavily penalizes extreme imbalance.
- **Intuition Example:**
- If a lazy model predicts "Positive" for every single person on Earth, its Recall is $100\%$, but its Precision is $0.001\%$.
- A simple average gives $\frac{100 + 0}{2} = 50\%$ (deceptive!).
- The **F1-Score** punishes the $0\%$ Precision and collapses to near **$0\%$**, correctly signaling that the model is useless.

## 5. Accuracy — "The Overall Scorecard"

$$
\text{Accuracy} = \frac{\text{True Positives} + \text{True Negatives}}{\text{All Cases}}
$$

- **Question it answers:** *"Out of every single attempt, how many total right answers did the model get?"*
- **The Fatal Trap of Accuracy (Class Imbalance):**
- Imagine a rare disease that only affects 1 in 1,000 people ($0.1\%$).
- A dummy model that simply predicts *"Everyone is Healthy"* every single time achieves **99.9% Accuracy**!
- It sounds amazing on paper, but it missed $100\%$ of sick patients ($\text{Recall} = 0\%$).

## Summary Cheat Sheet

| Metric | Primary Focus | Evaluates | Penalty Area |
| --- | --- | --- | --- |
| **Precision** | $\frac{\text{TP}}{\text{TP} + \text{FP}}$ | Quality of positive predictions | False Alarms ($\text{FP}$) |
| **Recall** | $\frac{\text{TP}}{\text{TP} + \text{FN}}$ | Coverage of real targets | Missed Targets ($\text{FN}$) |
| **Specificity** | $\frac{\text{TN}}{\text{TN} + \text{FP}}$ | Protection of negative cases | False Alarms ($\text{FP}$) |
| **F1-Score** | Harmonic Mean | Balance between Precision & Recall | Extreme imbalance in either |
| **Accuracy** | $\frac{\text{Correct}}{\text{Total}}$ | Overall correctness | Fails on imbalanced datasets |
